# Demo: split_dataset (Integrated)
This notebook cell demonstrates using demo_split.py added to the repository. It attempts to load the Spaceship Titanic dataset if available (as the original notebook does) and falls back to a small synthetic DataFrame otherwise.

Run the code cell below to see Copilot-friendly demo output.

```python
# Import the helper we added to the repo
from demo_split import split_dataset
import pandas as pd
import os

# Try to load the dataset the original notebook references; if not found, create a small demo DataFrame
dataset_path = '/kaggle/input/spaceship-titanic/train.csv'
if os.path.exists(dataset_path):
    df = pd.read_csv(dataset_path)
    print('Loaded dataset from', dataset_path)
else:
    df = pd.DataFrame({
        'feature': range(100),
        'label': [x % 2 for x in range(100)]
    })
    print('Using synthetic demo DataFrame (100 rows)')

# Run the split with a fixed seed for reproducibility
train, test = split_dataset(df, test_ratio=0.2, random_seed=42)
print(f'{len(train)} examples in training, {len(test)} examples in testing.')

# Show a small preview of the training set
train.head()
```
